```python
import os, uuid, hashlib, platform
# 1.填写学号（字符串！）
STUDENT_ID = "20244080112"

# 2.打印 conda 环境
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV", "base")
print(f"[Conda 环境] {CONDA_ENV}")

# 3.机器指纹
MACHINE_FP = hashlib.sha256((
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}"
).encode()).hexdigest()[:16]

# 4.生成唯一ID
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"

print(f"[学号]      {STUDENT_ID}")
print(f"[机器指纹]  {MACHINE_FP}")
print(f"[RUN_ID]    {RUN_ID}")
```

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
import math

def euler_limit(n):
    return (1.0 + 1.0 / n) ** n


def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
    return total


def euler_approx(eps):
    true_e = math.exp(1)
    total = 0.0
    k = 0
    fact = 1
    while abs(true_e - total) > eps:
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
        k += 1
    return total


def print_euler_sum_table(n):
    for k in range(1, n + 1):
        val = euler_infinite_sum(k)
        print(val)
```

```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True


def is_power(b, n):
    if n == 1:
        return True
    while n % b == 0:
        n = n // b
    return n == 1
```
