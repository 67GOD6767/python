```python
#1
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("out of range")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

#2 for求和
s_for = 0
for i in range(1,101):
    s_for += i
print("for sum", s_for)

s_while =0
i=1
while i<=100:
    s_while +=i
    i +=1
print("while sum", s_while)

#3 阶乘
def fact_iter(n):
    res =1
    for x in range(2,n+1):
        res *=x
    return res

def fact_rec(n):
    if n ==0:
        return 1
    return n * fact_rec(n-1)

print(fact_iter(0),fact_rec(0))
print(fact_iter(1),fact_rec(1))
print(fact_iter(5),fact_rec(5))
print(fact_iter(10),fact_rec(10))

#4 summary
def summary(values):
    minv = min(values)
    maxv = max(values)
    avg = sum(values)/len(values)
    return minv, maxv, avg

#5可变参数
def sum_all(*args):
    return sum(args)
```

```python
import os
import pickle

#1
print(os.getcwd())
data_dir = os.path.join(".","exp01_data")
os.makedirs(data_dir, exist_ok=True)

#2
text_data = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
file_path = os.path.join(data_dir,"students.txt")
with open(file_path,"w",encoding="utf-8") as f:
    f.write(text_data)

with open(file_path,"r",encoding="utf-8") as f:
    lines = f.readlines()[1:]

#3
def parse_student(line):
    parts = line.strip().split(',')
    if len(parts)!=2:
        raise ValueError("wrong field count")
    name,score_str = parts
    if len(name.strip())==0:
        raise ValueError("empty name")
    score = int(score_str)
    if not (0<=score<=100):
        raise ValueError("score out of [0,100]")
    return {"name":name.strip(),"score":score}

#4
valid = []
invalid_count = 0
for lineno,line in enumerate(lines, start=2):
    try:
        res = parse_student(line)
        valid.append(res)
    except Exception as err:
        invalid_count +=1

valid_count = len(valid)
avg_val = sum(x["score"] for x in valid)/valid_count
pass_count = sum(1 for s in valid if s["score"]>=60)
passrate = pass_count / valid_count

report_path = os.path.join(data_dir,"report.txt")
# w模式覆盖
with open(report_path,"w",encoding="utf-8") as f:
    f.write(f"有效记录:{valid_count}\n无效记录:{invalid_count}\n平均分:{avg_val:.2f}\n及格率:{passrate:.2%}\n")
# a追加模式
with open(report_path,"a",encoding="utf-8") as f:
    f.write("<<append test line>>\n")

#6 pickle
pk_path = os.path.join(data_dir,"students.pkl")
with open(pk_path,"wb") as fout:
    pickle.dump(valid,fout)
with open(pk_path,"rb") as fin:
    loaded_data = pickle.load(fin)
print(loaded_data == valid)

#8 file not found
try:
    with open("no_such_file.txt") as f:
        pass
except FileNotFoundError as e:
    print("捕获文件不存在异常：",e)

'''
#任务7 需要单独新建 student_utils.py 文件写入parse_student函数
# 在notebook里面调用：
from student_utils import parse_student
#测试
# parse_student("Tom,70") 有效
# parse_student("Bob,200") 无效抛异常
'''
```

```python
import copy

#1
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

#2
stu1 = Student("Alice",85)
print(stu1, stu1.is_passed())

#3 继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

#4深浅拷贝
origin = {"name":"Alice","scores":[85,90]}
d1 = origin
d2 = copy.copy(origin)
d3 = copy.deepcopy(origin)

origin["scores"][0] = 99
print("direct assign:",d1)
print("shallow copy:",d2)
print("deep copy:",d3)
```
