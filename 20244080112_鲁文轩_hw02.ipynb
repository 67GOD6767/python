```python
def is_palindrome(s):
    cleaned = [ch.lower() for ch in s if ch != ' ']
    return cleaned == cleaned[::-1]

def is_abecedarian(s):
    chars = [ch for ch in s if ch != ' ']
    for i in range(1, len(chars)):
        if chars[i] < chars[i-1]:
            return False
    return True

def remove_vowels(s):
    vowels = {'a','e','i','o','u'}
    res = []
    for ch in s:
        if ch.lower() not in vowels:
            res.append(ch)
    return ''.join(res)
```

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入不是列表类型")
    return lst[::-1]

def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i] < seq[i-1]:
            return False
    return True
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    for val in t:
        if not isinstance(val, (int, float)):
            raise TypeError("元组内元素必须是整数或浮点数")
    return tuple(x * s for x in t)

def vec_inner_product(x, y):
    if not isinstance(x, tuple):
        raise TypeError("第一个向量不是元组")
    if not isinstance(y, tuple):
        raise TypeError("第二个向量不是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度不一致")
    total = 0.0
    for xi, yi in zip(x, y):
        total += xi * yi
    return total

def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    row_len = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if row_len is None:
            row_len = len(row)
        else:
            if len(row) != row_len:
                return False
        for num in row:
            if not isinstance(num, (int, float)):
                return False
    return True

def mx_vec_mult(matrix, vec):
    if not check_valid_mx(matrix):
        raise ValueError("矩阵不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_cols = len(matrix[0])
    if len(vec) != n_cols:
        raise ValueError("向量维度和矩阵列数不匹配")
    result = []
    for row in matrix:
        s = 0.0
        for a, b in zip(row, vec):
            s += a * b
        result.append(s)
    return tuple(result)
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        self.dim = dim
        if elements is None:
            self.elements = [0.0 for _ in range(dim)]
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或者元组")
            if len(elements) != dim:
                raise ValueError("元素个数和维度不一致")
            for v in elements:
                if not isinstance(v, (int, float)):
                    raise TypeError("向量元素必须是整数或浮点数")
            self.elements = list(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同不能比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同不能比较")
        return self.elements < other.elements

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not (self <= other)

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法求内积")
        total = 0.0
        for a,b in zip(self.elements, other.elements):
            total += a*b
        return total

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_ele = [x * other for x in self.elements]
            return Vector(self.dim, new_ele)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量维度不一致")
            new_ele = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_ele)
        else:
            raise TypeError("乘法右侧必须是标量或者Vector")

    def __rmul__(self, other):
        return self * other
```
